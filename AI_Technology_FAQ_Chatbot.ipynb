```python
from difflib import get_close_matches

faq = {
    "what is ai": "Artificial Intelligence is the simulation of human intelligence in machines.",
    "what is machine learning": "Machine Learning is a branch of AI that learns from data.",
    "what is deep learning": "Deep Learning uses neural networks to solve complex problems.",
    "what is python": "Python is a popular programming language used in AI and data science.",
    "what is chatgpt": "ChatGPT is an AI chatbot developed by OpenAI.",
    "what is data science": "Data Science is the process of analyzing data to gain insights.",
    "what is nlp": "NLP stands for Natural Language Processing.",
    "what is computer vision": "Computer Vision helps computers understand images and videos.",
    "what is robotics": "Robotics combines engineering and AI to build intelligent machines.",
    "what is cloud computing": "Cloud Computing provides computing services over the internet.",
    "what is generative ai": "Generative AI creates new content such as text and images.",
    "what is automation": "Automation performs tasks automatically with minimal human effort."
}

print("🤖 AI & Technology FAQ Chatbot")
print("Type 'exit' to stop the chatbot.\n")

while True:
    question = input("You: ").lower()

    if question == "exit":
        print("Bot: Goodbye 👋")
        break

    match = get_close_matches(question, faq.keys(), n=1)

    if match:
        print("Bot:", faq[match[0]])
    else:
        print("Bot: Sorry, I don't understand your question.")
```
